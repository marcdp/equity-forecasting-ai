```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Datos de los eventos
events = [
    {"year": 1970, "label": "ARIMA\n(Box & Jenkins)", "color": "#1f77b4"},
    {"year": 1982, "label": "GARCH\n(Engle)", "color": "#1f77b4"},
    {"year": 1995, "label": "SVM\n(Vapnik)", "color": "#2ca02c"},
    {"year": 2001, "label": "Random Forest\n(Breiman)", "color": "#2ca02c"},
    {"year": 2014, "label": "Deep\nLearning", "color": "#d62728"},
    {"year": 2017, "label": "LSTM\nen\nfinanzas", "color": "#d62728"},
    {"year": 2018, "label": "Transformers", "color": "#d62728"},
    {"year": 2020, "label": "N-BEATS", "color": "#d62728"},
    {"year": 2023, "label": "Modelos\nhíbridos\n(TSFEDL)", "color": "#9467bd"},
]

# Crear la figura
fig, ax = plt.subplots(figsize=(14, 3))
y = 1  # Línea base

# Dibujar los eventos
for i, event in enumerate(events):
    ax.plot(event["year"], y, "o", color=event["color"], markersize=10)
    ax.text(event["year"], y + 0.1 if i % 2 == 0 else y - 0.1, event["label"],
            ha='center', va='bottom' if i % 2 == 0 else 'top', fontsize=9)

# Ajustes de la gráfica
ax.set_ylim(0.5, 1.5)
ax.set_xlim(1965, 2025)
ax.set_yticks([])
#ax.set_title("Línea de tiempo: Evolución de los modelos para predicción financiera", fontsize=14)
ax.set_xlabel("Año")
ax.grid(True, axis='x', linestyle='--', alpha=0.5)

# Leyenda
legend_patches = [
    mpatches.Patch(color="#1f77b4", label="Modelos tradicionales"),
    mpatches.Patch(color="#2ca02c", label="Modelos de ML"),
    mpatches.Patch(color="#d62728", label="Redes neuronales profundas"),
    mpatches.Patch(color="#9467bd", label="Modelos híbridos")
]
ax.legend(handles=legend_patches,  )

#plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Abril + Mayo + Junio + Julio
tasks = [
    {"Task": "Definición del problema y objetivos", "Start": "2025-04-15",       "End": "2025-05-01", "Hours": 10},
    {"Task": "Revisión del estado del arte",        "Start": "2025-05-01",       "End": "2025-06-10", "Hours": 30},
    {"Task": "Recolección y preparación de datos",  "Start": "2025-06-10",       "End": "2025-07-10", "Hours": 10},
    {"Task": "Implementación de los modelos",       "Start": "2025-07-10",       "End": "2025-09-10", "Hours": 25},
    {"Task": "Evaluación de los modelos",           "Start": "2025-08-10",       "End": "2025-09-15", "Hours": 20},
    {"Task": "Interpretación de resultados",        "Start": "2025-08-15",       "End": "2025-09-20", "Hours": 15},
    {"Task": "Redacción y revisión del documento final", "Start": "2025-09-20", "End": "2025-10-05", "Hours": 40},
]

df = pd.DataFrame(tasks)
df["Start"] = pd.to_datetime(df["Start"])
df["End"] = pd.to_datetime(df["End"])
df["Duration"] = (df["End"] - df["Start"]).dt.days

# plot
fig, ax = plt.subplots(figsize=(10, 5))

for i, row in df.iterrows():
    ax.barh(
        y=row["Task"],
        width=row["Duration"],
        left=row["Start"],
        color="#4A79A5",
        alpha=0.85
    )
    ax.text(
        row["End"],
        i,
        f' {row["Hours"]} h',
        va='center',
        ha='left',
        fontsize=9,
        color="black"
    )

# formatea
ax.set_xlabel("Fecha")
ax.set_ylabel("Fases del proyecto")
ax.invert_yaxis()
ax.grid(axis="x", linestyle="--", alpha=0.5)
fig.autofmt_xdate()
plt.xticks(rotation=0)
plt.show()

total_hours = df["Hours"].sum()
print(f"🔹 Total hours: {total_hours} h")
```
